```python
# --- Cell 2: environment + data ------------------------------------------
# Single GPU on purpose: DataParallel across T4 x2 adds a class of hangs that
# is not worth the modest speedup on a one-shot overnight run.
import os
os.environ["CUDA_VISIBLE_DEVICES"] = "0"

import torch
print("torch", torch.__version__, "| cuda", torch.cuda.is_available())
assert torch.cuda.is_available(), "No GPU. Settings -> Accelerator -> GPU T4 x2 / P100."
print(torch.cuda.get_device_properties(0).name)

from pathlib import Path
def find_dataset(name):
    root = Path("/kaggle/input")
    if (root / name).is_dir():
        return root / name
    for pat in (f"*/{name}", f"datasets/*/{name}"):
        m = [p for p in root.glob(pat) if p.is_dir()]
        if m:
            return m[0]
    raise FileNotFoundError(
        f"dataset '{name}' not attached. Present: {[p.name for p in root.iterdir()]}"
        if root.is_dir() else "/kaggle/input missing")

# Kaggle auto-extracts uploaded zips, so the source may be a file or a directory.
DATA_ROOT = find_dataset("kla-semicon-2026")
DATA_SOURCE = DATA_ROOT / "train.zip" if (DATA_ROOT / "train.zip").is_file() else DATA_ROOT
print("data source:", DATA_SOURCE)

import subprocess
subprocess.run(["pip", "-q", "install", "lpips"], check=False)  # optional extra

from src.data import pack_dataset
meta = pack_dataset(DATA_SOURCE, "/kaggle/working/packed")
```
